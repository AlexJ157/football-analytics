```python
import pandas as pd
df = pd.read_csv("../data/raw/Matches.csv")
df.head()
```

```python
df.columns.tolist()
```

```python
def get_season(date):
    if date.month >= 8:
        return f"{date.year}/{str(date.year + 1)[-2:]}"
    else:
        return f"{date.year - 1}/{str(date.year)[-2:]}"

df["MatchDate"] = pd.to_datetime(df["MatchDate"], dayfirst=True)
df["Season"] = df["MatchDate"].apply(get_season)
```

```python
prem_df = df[df["Division"] == "E0"]
prem_df = prem_df.sort_values("MatchDate").reset_index(drop=True)
prem_df.head()
```

```python
columns_to_keep = [
    "Season",
    "MatchDate",
    "HomeTeam",
    "AwayTeam",
    "Form3Home",
    "Form5Home",
    "Form3Away",
    "Form5Away",
    "FTHome",
    "FTAway",
    "FTResult"
]

prem_df = prem_df[columns_to_keep].copy()
prem_df.head()
```

```python
team_name_fixes = {
    "Nott'm Forest": "Nottingham Forest",
    "Nottm Forest": "Nottingham Forest"
}
prem_df["HomeTeam"] = prem_df["HomeTeam"].replace(team_name_fixes)
prem_df["AwayTeam"] = prem_df["AwayTeam"].replace(team_name_fixes)

sorted(prem_df["HomeTeam"].unique())
```

```python
prem_df.to_csv("../data/processed/clean_matches.csv", index=False)
```
