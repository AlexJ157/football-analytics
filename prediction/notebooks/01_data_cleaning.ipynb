```python
import pandas as pd
df = pd.read_csv("../data/raw/Matches.csv")
df.head()
```

```python
def get_season(date):
    if date.month >= 8:
        return f"{date.year}/{str(date.year + 1)[-2:]}"
    else:
        return f"{date.year - 1}/{str(date.year)[-2:]}"

df["MatchDate"] = pd.to_datetime(df["MatchDate"], dayfirst=True)
df["Season"] = df["MatchDate"].apply(get_season)
```

```python
seasons = [
    "2015/16",
    "2016/17",
    "2017/18",
    "2018/19",
    "2019/20",
    "2020/21",
    "2021/22",
    "2022/23",
    "2023/24",
    "2024/25"
]
df = df[df["Season"].isin(seasons)]
prem_df = df[df["Division"] == "E0"]
prem_df.head()
```

```python
columns_to_keep = [
    "MatchDate",
    "HomeTeam",
    "AwayTeam",
    "HomeElo",
    "AwayElo",
    "Form3Home",
    "Form5Home",
    "Form3Away",
    "Form5Away",
    "FTHome",
    "FTAway",
    "FTResult"
]

prem_df = prem_df[columns_to_keep].copy()
prem_df.head()
```

```python
prem_df.isnull().sum()
```

```python
prem_df = prem_df.dropna(subset=["HomeElo", "AwayElo"])
prem_df.isnull().sum()
```

```python
team_name_fixes = {
    "Nott'm Forest": "Nottm Forest",
}
prem_df["HomeTeam"] = prem_df["HomeTeam"].replace(team_name_fixes)
prem_df["AwayTeam"] = prem_df["AwayTeam"].replace(team_name_fixes)

sorted(prem_df["HomeTeam"].unique())
```
